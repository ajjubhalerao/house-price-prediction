## Feature Selection Discussion

The target variable for this project is `SalePrice`, which represents the selling price of a house.

Several features are likely to be useful predictors of house prices. Features related to house size, quality, location, age, and facilities are expected to have an important relationship with the selling price.

Some important potential predictors include:

- `OverallQual` – overall material and finish quality of the house.
- `GrLivArea` – above-ground living area.
- `TotalBsmtSF` – total basement area.
- `1stFlrSF` – first-floor area.
- `GarageCars` – number of cars that can be accommodated in the garage.
- `GarageArea` – garage size.
- `YearBuilt` – year the house was originally built.
- `FullBath` – number of full bathrooms.
- `BedroomAbvGr` – number of bedrooms above ground.
- `LotArea` – size of the property lot.
- `Neighborhood` – location of the property.

Both numerical and categorical features can provide useful information. Numerical features can be used directly after handling missing values, while categorical features such as `Neighborhood` and `MSZoning` will need to be converted into numerical form using One-Hot Encoding.

The final features will be determined after examining missing values and correlations with `SalePrice`.

```python
X = df.drop("SalePrice", axis=1)
y = df["SalePrice"]
```

```python
X = X.drop("Id", axis=1)
```

```python
X.shape
```

```python
numerical_features = X.select_dtypes(include=["int64", "float64"]).columns

print("Number of numerical features:", len(numerical_features))
```

```python
categorical_features = X.select_dtypes(include=["object"]).columns

print("Number of categorical features:", len(categorical_features))
```

```python
print("Numerical features:")
print(list(numerical_features))

print("\nCategorical features:")
print(list(categorical_features))
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
```

```python
numerical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median"))
])
```

```python
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_transformer, numerical_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("regressor", LinearRegression())
])
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred[:10]
```

```python
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np
```

```python
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("Mean Squared Error:", mse)
print("Root Mean Squared Error:", rmse)
print("R² Score:", r2)
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(x=y_test, y=y_pred)

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    "r--"
)

plt.title("Actual vs Predicted House Prices")
plt.xlabel("Actual Sale Price")
plt.ylabel("Predicted Sale Price")

plt.show()
```

```python
residuals = y_test - y_pred

plt.figure(figsize=(8, 6))

sns.scatterplot(x=y_pred, y=residuals)

plt.axhline(0, color="red", linestyle="--")

plt.title("Residual Plot")
plt.xlabel("Predicted Sale Price")
plt.ylabel("Residuals")

plt.show()
```

## Conclusion

This project focused on predicting house prices using Machine Learning and the Linear Regression algorithm. The Ames Housing dataset was used for this analysis, which contains information about different characteristics of residential properties and their corresponding sale prices. The main objective was to understand how different house features are related to property prices and to build a model that can predict the sale price of a house based on these features.

The project started with Exploratory Data Analysis (EDA), where the dataset was examined using its shape, column information, data types, missing-value analysis, and descriptive statistics. The target variable, `SalePrice`, was analyzed to understand its distribution. The distribution showed that house prices were right-skewed, meaning that most houses were concentrated in the lower-to-middle price range while a smaller number of houses had considerably higher prices.

Feature selection was then considered based on the meaning of the variables and their relationship with the target variable. Correlation analysis showed that features such as `OverallQual`, `GrLivArea`, `GarageCars`, `GarageArea`, `TotalBsmtSF`, `1stFlrSF`, `FullBath`, and `YearBuilt` had relatively strong relationships with `SalePrice`. `OverallQual` and `GrLivArea` showed particularly strong positive correlations with the target variable, indicating that overall house quality and above-ground living area are important factors associated with house prices in this dataset.

The dataset contained both numerical and categorical features, so appropriate preprocessing was required before training the machine learning model. Missing numerical values were handled using median imputation, while missing categorical values were handled using the most frequent value. Categorical variables were converted into numerical representations using One-Hot Encoding. The `Id` column was removed because it is an identifier and does not represent a meaningful property characteristic.

The data was divided into training and testing datasets using an 80/20 split. A preprocessing pipeline was created so that the required transformations could be applied consistently to the training and testing data. Linear Regression was then trained using the processed training data and used to predict house prices for the unseen test data.

The Linear Regression model achieved the following results:

- Mean Squared Error (MSE): **979,376,460.09**
- Root Mean Squared Error (RMSE): **31,294.99**
- R² Score: **0.8723**

The R² score of approximately **87.23%** indicates that the model explains a substantial portion of the variation in house sale prices within the test dataset. The RMSE of approximately **31,295** indicates the typical magnitude of prediction error in the same units as the target variable.

An Actual vs Predicted plot was created to visually compare the model's predictions with the actual house prices. Most observations were reasonably close to the perfect-prediction line, showing that the model was able to capture a strong relationship between the input features and house prices. A residual plot was also created to examine the prediction errors. The residuals were distributed around the zero line, although some larger errors were present, particularly for certain higher-priced properties.

Coefficient analysis was performed to understand the direction of relationships between the transformed features and the predicted house prices. Since categorical variables were converted using One-Hot Encoding, the model produced coefficients for individual categories such as neighborhoods, exterior materials, basement quality, and other property characteristics. Positive coefficients indicate a positive association with the predicted target within the fitted model, while negative coefficients indicate a negative association, holding the other model features constant.

As an additional experiment, Ridge Regression was implemented to compare its performance with the Linear Regression model. Using `alpha=1.0`, Ridge Regression produced:

- Mean Squared Error (MSE): **1,194,778,185.40**
- Root Mean Squared Error (RMSE): **34,565.56**
- R² Score: **0.8442**

The Ridge model was evaluated using the same test dataset, allowing the two approaches to be compared under the same conditions. The results demonstrate how different regression algorithms can produce different performance levels depending on the dataset, preprocessing, and model configuration.

Overall, this project demonstrates a complete end-to-end machine learning workflow for a regression problem. It includes data loading, exploratory data analysis, missing-value handling, feature selection, correlation analysis, categorical encoding, train-test splitting, pipeline construction, model training, prediction, evaluation, visualization, coefficient analysis, and comparison with a regularized regression model.

This project also provided practical experience in using important Python libraries such as **Pandas, NumPy, Matplotlib, Seaborn, and Scikit-learn**. It demonstrates how raw housing data can be transformed into a machine learning-ready format and used to build a predictive model. The results also show that house price prediction is influenced by multiple property characteristics, and machine learning can be used to identify relationships between these characteristics and sale prices.

Future improvements could include experimenting with additional regression algorithms such as Random Forest, Gradient Boosting, or XGBoost, applying feature engineering, tuning model hyperparameters, handling skewed numerical variables, and using cross-validation to obtain a more robust estimate of model performance. These improvements could potentially provide further insight into the factors affecting house prices and improve predictive performance.
